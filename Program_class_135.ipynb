```python
#1. Student
class Student:
    def __init__(self, roll_no, name, marks):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks

    def display(self):
        print("Roll No:", self.roll_no)
        print("Name:", self.name)
        print("Marks:", self.marks)
        print("Percentage:", sum(self.marks) / len(self.marks))

s1 = Student(1, "Rahul", [80, 75, 90])
s2 = Student(2, "Amit", [70, 85, 80])
s1.display()
s2.display()
```

```python
#2.Employee
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def calculate(self):
        hra = self.salary * 0.20
        da = self.salary * 0.10
        gross = self.salary + hra + da
        print("Employee ID:", self.emp_id)
        print("Name:", self.name)
        print("HRA:", hra)
        print("DA:", da)
        print("Gross Salary:", gross)

e1 = Employee(101, "Rahul", 30000)
e1.calculate()
```

```python
#3.Rectangle
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        print("Area:", self.length * self.breadth)

    def perimeter(self):
        print("Perimeter:", 2 * (self.length + self.breadth))

r = Rectangle(10, 5)
r.area()
r.perimeter()
```

```python
#4.Circle
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print("Area:", 3.14 * self.radius * self.radius)

    def circumference(self):
        print("Circumference:", 2 * 3.14 * self.radius)

c = Circle(5)
c.area()
c.circumference()
```

```python
#5.Book
class Book:
    def __init__(self, book_id, title, author, price):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.price = price

    def display(self):
        print(self.book_id, self.title, self.author, self.price)

b1 = Book(1, "Python", "ABC", 500)
b2 = Book(2, "Java", "XYZ", 450)
b3 = Book(3, "C++", "PQR", 400)
b1.display()
b2.display()
b3.display()
```

```python
#6.Electricity_Bill
class ElectricityBill:
    def __init__(self, number, name, units):
        self.number = number
        self.name = name
        self.units = units

    def calculate(self):
        if self.units <= 100:
            bill = self.units * 5
        elif self.units <= 200:
            bill = self.units * 7
        else:
            bill = self.units * 10
        print("Consumer:", self.name)
        print("Bill:", bill)

b = ElectricityBill(101, "Rahul", 150)
b.calculate()
```

```python
#7.Mobile_Phone
class MobilePhone:
    def __init__(self, brand, model, storage, price):
        self.brand = brand
        self.model = model
        self.storage = storage
        self.price = price

    def display(self):
        print(self.brand, self.model, self.storage, self.price)

    def discount(self, percent):
        final = self.price - (self.price * percent / 100)
        print("Price after discount:", final)

m = MobilePhone("Samsung", "S24", "128GB", 50000)
m.display()
m.discount(10)
```

```python
#8.Patient
class Patient:
    def __init__(self, pid, name, age, disease, fee):
        self.pid = pid
        self.name = name
        self.age = age
        self.disease = disease
        self.fee = fee

    def display(self):
        print("ID:", self.pid)
        print("Name:", self.name)
        print("Age:", self.age)
        print("Disease:", self.disease)

    def bill(self):
        print("Total Bill:", self.fee)

p = Patient(1, "Rahul", 25, "Fever", 500)
p.display()
p.bill()
```

```python
#9ATM
class ATM:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def check(self):
        print("Balance:", self.balance)

    def deposit(self, amount):
        self.balance += amount
        print("Money Deposited")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Money Withdrawn")
        else:
            print("Insufficient Balance")

    def details(self):
        print("Name:", self.name)
        print("Balance:", self.balance)

a = ATM("Rahul", 5000)

while True:
    print("\n1.Check Balance")
    print("2.Deposit")
    print("3.Withdraw")
    print("4.Account Details")
    print("5.Exit")
    ch = int(input("Enter choice: "))
    if ch == 1:
        a.check()
    elif ch == 2:
        a.deposit(int(input("Enter amount: ")))
    elif ch == 3:
        a.withdraw(int(input("Enter amount: ")))
    elif ch == 4:
        a.details()
    elif ch == 5:
        break
```

```python
#10. Vechicle
class Vehicle:
    def __init__(self, number, model, rate):
        self.number = number
        self.model = model
        self.rate = rate
        self.available = True

    def rent(self, days):
        if self.available:
            self.available = False
            print("Vehicle Rented")
            print("Charges:", self.rate * days)
        else:
            print("Not Available")

    def return_vehicle(self):
        self.available = True
        print("Vehicle Returned")

v = Vehicle("MH09AB1234", "Swift", 1000)
v.rent(3)
v.return_vehicle()
```

```python
#11. Shopping_Cart
class ShoppingCart:
    def __init__(self, name, cart_id):
        self.name = name
        self.cart_id = cart_id
        self.products = []

    def add(self, product, price):
        self.products.append([product, price])

    def remove(self, product):
        for p in self.products:
            if p[0] == product:
                self.products.remove(p)
                break

    def total(self):
        total = 0
        for p in self.products:
            total += p[1]
        print("Total Bill:", total)

    def __del__(self):
        print("Shopping Cart Deleted")

c = ShoppingCart("Rahul", 101)
c.add("Pen", 20)
c.add("Book", 100)
c.remove("Pen")
c.total()
del c
```

```python
#12. Food Order
class FoodOrder:
    def __init__(self, order_id, name, item, quantity, price):
        self.order_id = order_id
        self.name = name
        self.item = item
        self.quantity = quantity
        self.price = price

    def bill(self):
        total = self.quantity * self.price
        tax = total * 0.05
        print("Total Bill:", total + tax)

    def __del__(self):
        print("Order Completed")

o = FoodOrder(1, "Rahul", "Pizza", 2, 200)
o.bill()

```

```python
#13. Student_Result
class StudentResult:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def result(self):
        total = sum(self.marks)
        percentage = total / 5
        if percentage >= 75:
            grade = "A"
        elif percentage >= 60:
            grade = "B"
        elif percentage >= 40:
            grade = "C"
        else:
            grade = "Fail"
        print("Name:", self.name)
        print("Total:", total)
        print("Percentage:", percentage)
        print("Grade:", grade)

    def __del__(self):
        print("Student Result Completed")

s = StudentResult("Rahul", [80, 75, 70, 85, 90])
s.result()
del s
```
